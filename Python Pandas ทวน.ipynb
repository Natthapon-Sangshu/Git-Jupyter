## การเปรียบเทียบค่า และตรรกศาสตร์

```python
## == เท่ากับ
## > มากกว่า  < น้อยกว่า  
##  >=  มากกว่าเท่ากับ  <=  น้อยกว่าเท่ากับ    !=ไม่เท่ากับ
## and     หากมี False เป็น False  
## or      True ตัวใดตัวนึง True    
## not     ตรงกันข้าม not true ก็ได้ False ในทิศทางตรงข้าม
x,y = 5,3
z,n = 3,10

print(x > y)
print(x < y)
print(y == z)
print(y >= z)
print(y <= z)
print(x <= n)
print(x != n)
print('ตรรกศาตร์ 1 ',x > y and z < n)
print('ตรรกศาตร์ 2 ',x > y or z/x < n)
print('ตรรกศาตร์ 3 ',y > x or x/y > y)
print('ตรรกศาตร์ 4 ',not(y > x and x/y > y))
```

## Recursive Function เรียกตัวมันเองอีกรอบในฟังก์ชั่น

```python
def fract(n=0):
    if n == 0:
        return 1
    else :
        return n*fract(n-1)         ##  พอมาถึง return ตัวนี้จะไปเรียกฟังก์ชั่นตัวเองใหม่อีกรอบ 
fract(5)

## รอบแรก n คือ 5* (fract(4) ---> 4* fract(3) ----> 3*fract(2)----> 2*fract(1) ----1*fract(0) ---->return 1) ก้อนฟังก์ชั่นที่ซ้อนอยู่
## ส่งค่าย้อนกลับ 
## fract 0 = 1    1*1
## fract 1 = 1    1*1
## fract 2 = 2    2*1
## fract 3 = 6    3*2
## fract 4 = 24    4*6
## fract 5 = 120    5*24
```

```python
## คำนวนลำดับ  ลำดับฟีโบนัชชี (Fibonacci sequence)
def somo(n):
    if   n == 1:
        return 1
    elif n == 2:
        return 2
    else :
        return somo(n-1) +somo(n-2)
## somo(5)
somo(10)

## somo(5) = somo(4) + somo(3) = 5+3 = 8
## somo(4) = somo(3) + somo(2) = 3+ 2 = 5  
## somo(3) = somo(2) + somo(1) = 2 + 1 = 3
## somo(2) = 2
## somo(1) = 1
## ต้องไล่ย้อนกลับจาก somo(1)
```

## List comprehension

```python
num = [1,5,7,9,6]
newlist= []           ## สร้างตัวแปร newlist เพื่อเก็บค่า

for x in num:
    newlist.append(x+5)
print(newlist)  

## List comprehension
num1 = [1,5,7,9,6]
newlist1 = [x+5 for x in num]    ## เอา x+5 มาใน newlist1 เลย และยัด for x in num มาด้วย
newlist2 = [x+5 for x in num if x !=6 ]
newlist3 = [x+5 if x < 6 else x-1 for x in num]  ## เขียนสลับที่ ความหมายเปลี่ยน  
my_dic = {x:x*2 for x in num}                    ## ลองใน Dic
print(newlist1)
print(newlist2)
print(newlist3)
print(my_dic)
```

```python
## map เอาฟังก์ชั่นที่วิ่งไป map จับกับ list
list1 = [7,8,3,9]
def test(x):
    return x*2
map1 = map(test,list1)
## map1 = map(lambda x: x*2, list1)     ## map(lambda ตัวแปร : operation ให้ทำอะไร ,ตัวแปรที่จะให้ทำ) 
## print(map1)                       ## แบบนี้จะแสดง object
print(list(map1))
```

```python
## filter จะเก็บค่าที่ true เท่านั้น
list1 = [7,8,3,9]
def test(x):
    return x/2 > 3
filter1 = filter(test,list1)
## filter1 = map(lambda x: x/2 > 3, list1)     ## map(lambda ตัวแปร : operation ให้ทำอะไร ,ตัวแปรที่จะให้ทำ) 
## print(map1)                       ## แบบนี้จะแสดง object
print(list(filter1))
```

## การเขียนโปรแกรมเชิงวัตถุ  Object Oriented Programming(OOP)

```python
## Class    Object   Properties   Methods
## สมมุติตัวอย่าง
## Class คือ person
## object คือ name
## properties คือ Email address age
## methods  คือ send email  change address
## class คือ จัดกลุ่มโค้ดที่เกี่ยวข้องกัน ให้เป็นระเบียบ เช่น **ฟังก์ชันหลาย ๆ อันที่ทำงานในเรื่องเดียวกัน** ไม่จำเป็นต้องมีครบ ทั้ง object properties 


class Person:                           ## ชื่อ class
    def __init__(self,name,age):         ## Object ต้องมีขีดล่างหน้า 2 อัน init ขีดล่างหลัง 2 อัน ไม่งั้น error  โดย Properties คือ name age alive
        self.name = name                ## ตัวแปรคือ self ในการวิ่ง เก็บ 2 ค่า name age ส่วน alive คือ defalut
        self.age = age
        self.alive = 'alive'

    def hello(self):                          ## Method hello
        print('Hello,my name is ' + self.name + ', I am '+ str(self.age) + ' years old.')


person1 = Person('John',30)
print(person1.name)
print(person1.age)
print(person1.alive)
person1.hello()
```

## การสืบทอด class   Inheritances 

```python
## Class ใหม่ชื่อ student ไปดึง Object propetries methods มาจาก Class person ทื่เคยสร้างไว้

class Person:                           ## ชื่อ class
    def __init__(self,name,age):         ## Object ต้องมีขีดล่างหน้า 2 อัน init ขีดล่างหลัง 2 อัน ไม่งั้น error  โดย Properties คือ name age alive
        self.name = name                ## ตัวแปรคือ self ในการวิ่ง เก็บ 2 ค่า name age ส่วน alive คือ defalut
        self.age = age
        self.alive = 'alive'

    def hello(self):                          ## Method hello
        print('Hello,my name is ' + self.name + ', I am '+ str(self.age) + ' years old.')

class Student(Person):                       ## () จะใส่ชื่อ class ที่ไปดึง
    def __init__(self,name,age,student_id):  ## Object ที่เก็บค่าของ class นี้
        super().__init__(name,age)           ## ดึง name age จาก class ก่อนด้วย super().
        self.student_id = student_id        ## สร้างตัวแปรรับค่าเฉพาะส่วนที่เพิ่ม

    def hellostudent(self):
        print('Hello,my name is ' + self.name +' is studying.' + 'I am ' + str(self.age) + ' years old. รหัสนักศึกษา :' + str(self.student_id))

SomO = Student('SomO',30,'01474')
SomO.hellostudent()
        
```

## WORKSHOP สร้างรายการซื้อของลูกค้า

```python
## สร้าง bill ซื้อของลูกค้า โดยสร้าง Object ลูกค้รา แสดง list สินค้าที่ซื้อ และราคารวม
## ถ้าเป็นสมาช้ิกของร้าน ลด 5%
## product = {'ขนม': 10,'เครื่องดื่ม':20,'แชมพู': 30,'ครีมทาผิว':40}

products = {'ขนม': 10,'เครื่องดื่ม':20,'แชมพู': 30,'ครีมทาผิว':40}

class Bill:
    def __init__(self,member = False): ## ไม่ต้องเติม bought เป็น list เปล่าๆ เก็บค่า
        self.member = member
        self.bought = []               ## สร้างตัวแปร bought เป็น list เปล่าๆ สำหรับเก็บค่า เก็บข้อมูลทั้งหมดที่ลูกค้าซื้อว่าชนิดไหนบ้าง
 
    def buy(self,products,price):            ## ฟังก์ชั่น เพิ่มรายการสินค้าที่ซื้อ เก็บ 2 ค่าเข้า list คือ 'ขนม': 10 สินค้า และราคา
        self.bought.append((products,price))  ## append   ก็คือนำค่ามาต่อกัน
      
    def show_bill(self):               ## ฟังก์ชั่น เมธอดการแสดงบิล
        price = 0                      ## เริ่มที่ Price = 0 เหมือนตอนบิลเริ่มต้นคือ 0 บาท
        for p in self.bought:                      ## ใช้ For loop วน ใน self.bought ว่าซื้อกี่รายการ
            print(f'{p[0]}................{p[1]}  บาท')    ## p[0] ตำแหน่ง index แรก คือ ขนม  p[1] ตำแหน่ง index 1  ราคา
            price += p[1]                                  ## เก็บค่ารวมของราคา +=
        print(f'รวม.....................{price} บาท')
        if self.member:
            print(f'ส่วนลด.................{price*0.05}  บาท')
            print(f'รวม..............{price-price*0.05}  บาท')

    ## def buy(self,products,price):            ## ฟังก์ชั่น เพิ่มรายการสินค้าที่ซื้อ
        ## self.bought.append((products,price))  ## append

Bill1 = Bill(True)  ## Biil1 = เรียกใช้ฟังก์ชั่น Bill เรื่องเป็นสมาชิก  True /Faluse

p = 'ขนม'                      ## รายการแรก ชื่อสินค้า ขนม ที่ต้องตั้งเป็น p เพราะมันจะไป link กับฟังก์ชั่นที่ 3
Bill1.buy(p,products[p]) ## Biil1 = เรียกใช้ฟังก์ชั่น buy เพิ่มรายการซื้อเก็บค่า 2 ตัว  p คือชื่อสินค้า ลงตรง product'ขนม' และ products['ขนม'] ลงตรง price

p = 'เครื่องดื่ม'                  ## รายการที่เพิ่ม ในฟังก์ชั่น buy มี .append จึงเติมค่าตลอดไม่ซ้ำกัน
Bill1.buy(p,products[p])

p = 'ขนม'                    ## รายการที่เพิ่ม ในฟังก์ชั่น buy มี .append จึงเติมค่าตลอดไม่ซ้ำกัน
Bill1.buy(p,products[p])

p = 'ครีมทาผิว'                    ## รายการที่เพิ่ม ในฟังก์ชั่น buy มี .append จึงเติมค่าตลอดไม่ซ้ำกัน
Bill1.buy(p,products[p])


Bill1.show_bill()   ## Biil1 = เรียกใช้ฟังก์ชั่น show_bill
```
