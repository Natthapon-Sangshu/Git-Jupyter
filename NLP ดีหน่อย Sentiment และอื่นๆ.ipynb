```python
##ฟังก์ชั่นจัดการเครียร์ทุกสิ่งอย่างพร้อมเข้า Model

import re
import unicodedata
import emoji
from bs4 import BeautifulSoup
from pythainlp import word_tokenize


def cleanfor_transformer(getdataframe):
    output = getdataframe.str.strip()
    output = output.apply(lambda x: unicodedata.normalize("NFC", x))
    
    # Remove HTML tags
    output = output.apply(lambda x: BeautifulSoup(x, "html.parser").get_text())
    
    # Remove URLs
    output = output.str.replace(r'http\S+|www\S+', '', regex=True)
    
    # Remove emails
    output = output.str.replace(r'\S+@\S+', '', regex=True)
    
    # Remove non-text except Thai/English numbers
    output = output.str.replace(r'[^\w\sก-๙]', ' ', regex=True)
    
    # Reduce multiple spaces
    output = output.str.replace(r'\s+', ' ', regex=True)
    
    # Reduce repeated characters
    output = output.str.replace(r'(.)\1{2,}', r'\1\1', regex=True)
    
    # Remove emoji
    output = output.apply(lambda x: emoji.replace_emoji(x, ''))
    
    # ตัดคำด้วย attacut 
    output = output.apply(
    lambda x: " ".join(word_tokenize(x, engine="attacut"))
)
   
    return output

```

```python
def get_similarmovies(embeddings,movie_index,movie_details,top_n=3):
    # สร้าง Movie embedding สำหรับ movie index
    m_embedding = np.array(embeddings[movie_index]).reshape(1,-1)
    # คำนวน similarity score
    
    similarity_score = cosine_similarity(m_embedding, movie_embeddings)[0]
    similarity_score_series = pd.Series(similarity_score.flatten(),name='similarity_score')

    #bring in movie info
    movies_similarity_score_df = pd.concat([movie_details,similarity_score_series],axis=1)

    # นำมาใน movie info
    return movies_similarity_score_df.sort_values('similarity_score',ascending=False).iloc[0:top_n+1]
```

```python
import pandas as pd
import numpy as np
from transformers import pipeline
from sklearn.metrics.pairwise import cosine_similarity

model = 'sentence-transformers/all-MiniLM-L6-v2'
feature_extraction = pipeline('feature-extraction',model = model,device=-1)

book_df = pd.read_csv(r'D:\เรียน Data Science\NLP แบบดีหน่อย\Natural+Language+Processing+in+Python\Course Materials\Data\childrens_books.csv')

embeddings = book_df['Description'].apply(lambda x: np.mean(feature_extraction(x)[0], axis=0))
embeddings_book = np.vstack(embeddings)
embeddings_book.shape
```

```python
def get_similarbook(embeddings,book_index,description,top_n=3):

    # ระบุ สำหรับ book 
    m_embedding = np.array(embeddings[book_index]).reshape(1,-1)

    # คำนวน similarity score
    
    similarity_score = cosine_similarity(m_embedding, embeddings_book)[0]
    similarity_score_series = pd.Series(similarity_score.flatten(),name='similarity_score')

    # นำมาใน movie info
    book_similarity_score_df = pd.concat([description,similarity_score_series],axis=1)

    # Display movie recs
    return book_similarity_score_df.sort_values('similarity_score',ascending=False).iloc[0:top_n+1]
```
